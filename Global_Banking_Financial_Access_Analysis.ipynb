# **Data Cleaning**

```python
# Drop unnecessary columns

df.drop(columns=[
    "STRUCTURE",
    "ACTION",
    "FREQ",
    "INDICATOR",
    "SEX",
    "FINANCIAL_SECTOR",
    "ACCOUNT_ENTRY",
    "UNIT_MEASURE",
    "AGG_METHOD",
    "UNIT_TYPE",
    "DECIMALS",
    "DATABASE_ID",
    "TIME_FORMAT",
    "UNIT_MULT",
    "OBS_STATUS",
    "OBS_CONF",
    "SEX_LABEL",
    "ACCOUNT_ENTRY_LABEL",
    "AGG_METHOD_LABEL",
    "UNIT_TYPE_LABEL",
    "DECIMALS_LABEL",
    "TIME_FORMAT_LABEL",
    "UNIT_MULT_LABEL",
    "OBS_CONF_LABEL"
], inplace=True)

print(df.columns)
```

```python
# Rename columns
df.columns = [
    "Structure_ID",
    "Country_Code",
    "Year",
    "Value",
    "Frequency",
    "Country",
    "Indicator",
    "Financial_Sector_Label",
    "Unit_Measure",
    "Data_Source",
    "Status"
]
```

```python
# Checking unique values all cloumns

print("Structure_ID:", df["Structure_ID"].nunique())
print("Country_Code:", df["Country_Code"].nunique())
print("Year:", df["Year"].nunique())
print("Value:", df["Value"].nunique())
print("Frequency:", df["Frequency"].nunique())
print("Country:", df["Country"].nunique())
print("Indicator:", df["Indicator"].nunique())
print("Financial_Sector_Label:", df["Financial_Sector_Label"].nunique())
print("Unit_Measure:", df["Unit_Measure"].nunique())
print("Data_Source:", df["Data_Source"].nunique())
print("Status:", df["Status"].nunique())
```

```python
# Convert first letter of each word to uppercase

df["Indicator"] = df["Indicator"].str.strip().str.title()

df["Status"] = df["Status"].str.strip().str.title()
```

```python
# Check Nulls
df.isnull().sum()
```

```python
# Remove Null Value
df["Value"] = df["Value"].fillna(df["Value"].median())
```

```python
# Recheck the null value
print(df["Value"].isnull().sum())
```

## **Future Engineering**

```python
# adding Future Engineering
# Year_Category
df["Year_Category"] = df["Year"].apply(
    lambda x: "Old" if x < 2020 else "Recent"
)
```

```python
# Banking Level
df["Banking_Level"] = pd.cut(
    df["Value"],
    bins=[0,10,100,1000,100000],
    labels=["Low","Medium","High","Very High"]
)
```

```python
# Latest_Year
latest_year = df["Year"].max()

df["Latest_Data"] = df["Year"].apply(
    lambda x: "Yes" if x == latest_year else "No"
)
```

```python
# Banking Performance
df["Performance"] = np.where(
    df["Value"] >= df["Value"].mean(),
    "Above Average",
    "Below Average"
)
```

```python
# check the null values of the columns selected for future engineering
df.isnull().sum()
```

```python
## Handling the missing values of the columns selected for future engineering
df["Banking_Level"] = df["Banking_Level"].astype("object").fillna("Unknown")
```

```python
# check the dtypes of the columns selected for future engineering
df.dtypes
```

```python
# Data Type Conversion
df['Banking_Level'] = df['Banking_Level'].astype('object')
```

```python
df.reset_index(drop=True, inplace=True)
```

# **Visuvalization**

### **UNIVARIATE ANALYSIS**

```python
# A. Histogram
plt.figure(figsize=(10,6))
plt.hist(df["Value"], bins=10, color="skyblue", edgecolor="black")
plt.title("Distribution of Banking Values")
plt.xlabel("Banking Value")
plt.ylabel("Frequency")
plt.show()

# Purpose:
# Data Distribution
# Identify Skewness
```

### **Interpretation**
**Distribution of banking values**

1. Most banking values are concentrated near 0, indicating that the majority of records have relatively low banking values.

2. Only a few records contain extremely high banking values, extending up to approximately 1.5 × 10¹⁷.

3. The distribution is positively skewed (right-skewed) because the data is concentrated on the left with a long tail toward higher values.

4. The presence of these extreme values suggests that a small number of countries or records have exceptionally high banking values compared to the rest.

```python
## B.Log Scale Histogram
plt.figure(figsize=(10,6))

plt.hist(np.log1p(df["Value"]), bins=30,
         color="skyblue",
         edgecolor="black")

plt.title("Log Distribution of Banking Values")
plt.xlabel("Log(Value)")
plt.ylabel("Frequency")

plt.show()
```

# **Interpretation**

**Log Distribution of Banking Value**

1. Most banking values are concentrated between Log(Value) 5 and 8, indicating that a large number of records fall within this range.

2. The frequency gradually decreases as the log value increases, showing that higher banking values are less common.

3. A small number of records have very high banking values, extending up to approximately Log(Value) 40.

4. The log transformation reduces the effect of extreme values and provides a clearer view of the overall distribution.

```python
# C. Count Plot (Top 10 Countries)
top10 = df["Country"].value_counts().head(10)

plt.figure(figsize=(10,5))
plt.bar(top10.index,
        top10.values,
        color="skyblue",
        edgecolor="black")

plt.xticks(rotation=45)
plt.title("Top 10 Countries by Number of Records")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()

# Purpose
# Count Records
# Compare Countries
```

## **Interpretation**
**Top 10 Countries by Number of Records**

1. Bangladesh and Pakistan have the highest number of records, with approximately 1,460 records each.

2. Guinea follows closely with around 1,420 records, while Albania, Mauritius, Seychelles, and Madagascar each have approximately 1,320–1,350 records.

3. Botswana, Georgia, and Comoros have slightly fewer records, ranging from approximately 1,240 to 1,300.

4. The difference in record counts among the top 10 countries is relatively small, indicating that the dataset has a fairly balanced representation for these countries.


### **BIVARIATE ANALYSIS**

```python
# A. Year vs Average Banking Value
year_avg = df.groupby("Year")["Value"].mean()

plt.figure(figsize=(10,5))

plt.plot(year_avg.index,
         year_avg.values,
         marker='o',
         linewidth=2)

plt.title("Average Banking Value by Year")
plt.xlabel("Year")
plt.ylabel("Average Value")
plt.grid(True)

plt.show()

# Purpose
# Trend Analysis
# Compare Years
```

## **Interpretation**

**Average Value**

1. The average banking value increased gradually from approximately 8.0 × 10¹² in 2017 to 1.08 × 10¹³ in 2019.

2. There was a slight decrease in 2020 to around 1.03 × 10¹³, followed by a steady increase from 2021 to 2024.

3. The highest average banking value was recorded in 2024, reaching approximately 2.67 × 10¹³.

4. In 2025, the average banking value dropped sharply to nearly 0, indicating that the data for 2025 may be incomplete or only partially available.

```python
# B. Country vs Average Value
plt.figure(figsize=(10,6))

plt.hlines(y=top10.index,
           xmin=0,
           xmax=top10.values,
           color="skyblue")

plt.plot(top10.values,
         top10.index,
         "o",
         color="blue")

plt.title("Top 10 Countries by Average Banking Value")
plt.xlabel("Average Value")
plt.ylabel("Country")

plt.show()
```

### **Interpretation**

**Top 10 Countries by Average Banking Value**

1. Vietnam has the highest average banking value at approximately 1.24 × 10¹⁵.

2. Iran, Islamic Rep. ranks second with an average value of about 8.0 × 10¹⁴.

3. Indonesia and Korea, Rep. also show relatively high banking values compared to the remaining countries.

4. The remaining countries have considerably lower average banking values, indicating differences in banking performance.

```python
# C. Financial Sector vs Total Value
sector = df.groupby("Financial_Sector_Label")["Value"].sum().sort_values()

# Highlight highest value
colors = ["skyblue"] * len(sector)
colors[-1] = "orange"   # Highest value

plt.figure(figsize=(10,8))

plt.barh(
    sector.index,
    sector.values,
    color=colors,
    edgecolor="black"
)

plt.title("Total Banking Value by Financial Sector")
plt.xlabel("Total Banking Value")
plt.ylabel("Financial Sector")

plt.show()

# Purpose
# Sector Analysis
# Business Comparison
```

## **Interpretation**

**Financial Sector vs Total Banking Value**

1. Commercial Banks recorded the highest total banking value at approximately 1.68 × 10¹⁸.

2. Not Applicable recorded around 0.03 × 10¹⁸, which is much lower than Commercial Banks.

3. Credit Unions & Credit Cooperatives contributed approximately 0.02 × 10¹⁸.

4. Other Deposit Takers and Microfinance Institutions each recorded less than 0.01 × 10¹⁸, indicating a very small contribution



### **3. MULTIVARIATE ANALYSIS**

```python
# Count Banking Level by Year
line_data = df.groupby(['Year', 'Banking_Level']).size().unstack(fill_value=0)

# Create Line Chart
plt.figure(figsize=(10,6))

for column in line_data.columns:
    plt.plot(line_data.index, line_data[column], marker='o', linewidth=2, label=column)

plt.title("Banking Level Trend by Year")
plt.xlabel("Year")
plt.ylabel("Count")

# Legend outside the chart
plt.legend(title="Banking Level",
           loc="center left",
           bbox_to_anchor=(1.02, 0.5))

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()




```

## **Interpretation**

**Banking Level Distribution by Year**

1. Very High banking level remained the highest, increasing from ≈5,100 (2017) to ≈5,300 (2024).

2. Unknown banking level stayed stable at around ≈4,000 throughout the years.

3. Medium banking level increased slightly from ≈2,300 to ≈2,450.

4. High banking level rose from ≈1,500 to ≈1,700, while Low banking level decreased slightly from ≈1,900 to ≈1,800.

```python
# Country + Performance

# Count by Country
country_count = df['Country'].value_counts().head(10)

plt.figure(figsize=(10,6))

plt.barh(
    country_count.index,
    country_count.values,
    color='skyblue',
    edgecolor='black'
)

plt.title("Top 10 Countries by Record Count", fontsize=14)
plt.xlabel("Count")
plt.ylabel("Country")

plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
